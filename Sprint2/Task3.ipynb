```
//Мой итог

public interface ICashbackable // Интерфейс для кэшбэка
{
    decimal CashbackPercent { get; }
    decimal CalculateCashback(decimal purchaseAmount);
}

public interface IInsurable // Страхуемый   // Интерфейс для страхования
{
    bool HasInsurance { get; set; }
    void ActivateInsurance();      
}

public interface ICreditLimit  // Интерфейс для кредитного лимита
{
    decimal CreditLimit { get; }
    decimal AvailableCredit { get; }
    void IncreaseCreditLimit(decimal amount);        
}


public abstract class BonusCreditCard : CreditCard, ICashbackable   //Простое наследование: Один базовый класс + один интерфейс
{ 
    public int BonusPoints { get; protected set; }                                         
    public decimal CashbackPercent { get; protected set; }                                    

    public BonusCreditCard(string _cardNumber, string _holderName, string _expiryDate, decimal _cashbackPercent)
        : base(_cardNumber, _holderName, _expiryDate)
    {
        CashbackPercent = _cashbackPercent;
        BonusPoints = 0;
    }                                     
    public virtual void AddBonusPoints(int points)
    { 
        BonusPoints += points;
        Console.WriteLine($"Начислено {points} бонусных баллов. Всего: {BonusPoints}");   
    }                                
    public virtual decimal CalculateCashback(decimal purchaseAmount)
    { 
        return purchaseAmount * CashbackPercent / 100;
    }  
}                                                                                           

//  Базовому классу CreditCard: • Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок 
// действия (ExpiryDate).
public class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    public decimal Balance { get; protected set; }
    // НОВЫЕ атрибуты 
    public string Currency { get; set; }           //  НОВЫЙ  Валюта
    public bool IsActive { get; set; }             //  НОВЫЙ 

public CreditCard(string _cardNumber, string _holderName, string _expiryDate, decimal _balance = 1000, string _currency = "RUB" )
{
    CardNumber = _cardNumber;
    HolderName = _holderName;
    ExpiryDate = _expiryDate;
    Balance = _balance;
    Currency = _currency;     //Валюта         //  Инициализация нового атрибута
    IsActive = true;                           //  Инициализация нового атрибута
}
public static int TotalCard{ get; private set; }

    public CreditCard()
    {
        TotalCard++;
    }
    public static void Info()
    {
        Console.WriteLine($"Всего кредитных карт {TotalCard}");
    }
// • Методы: 
// НОВЫЕ методы 
    public virtual void ActivateCard()             //  НОВЫЙ МЕТОД
    {
        IsActive = true;
        Console.WriteLine($"Карта {CardNumber} активирована");
    }

    public virtual void BlockCard()                //  НОВЫЙ МЕТОД  
    {
        IsActive = false;
        Console.WriteLine($"Карта {CardNumber} заблокирована");
    }
// o GetInfo(): метод для получения информации о кредитной карте в виде строки.
    public virtual string GetInfo()
    {
        return $"Номер карты ({CardNumber}), Холдера {HolderName}, Срок действия: {ExpiryDate}";
    }
 
// o Pay(): метод для оплаты покупки с использованием карты. 
    public virtual void Pay(decimal amound)
    {
        Console.WriteLine($" Попытка оплаты на сумму {amound:C} с карты ({CardNumber}) ");
    }

// o CheckBalance(): метод для проверки баланса на карте. 
    public virtual void CheckBalance()
    {
        Console.WriteLine($" Проверка баланса карты ({CardNumber}), на ней {Balance:C}  ");       //Balance:C  :С это округление   как в России
    } 
}



// Требования к производным классам: 
// 1. GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты, 
// такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть 
// переопределен для добавления информации о получении бонусных миль 
// при оплате покупки. 
public class GoldCreditCard : BonusCreditCard, IInsurable //Сложное наследование: → CreditCard → BonusCreditCard → GoldCreditCard (3 уровня иерархии)
{
    public int BonusMiles { get; set; }
     // НОВЫЕ атрибуты 
    public bool HasLoungeAccess { get; set; }      //НОВЫЙ Имеет доступ в зал ожидания 
    
public GoldCreditCard(string _cardNumber, string _holderName, string _expiryDate )// ((, decimal _balance = 1000 )  _balance  )
                    :base(_cardNumber, _holderName, _expiryDate, 3.0m) 
    {
        BonusMiles = 0;
        HasLoungeAccess = true;                    // (Новый атрибут) Имеет доступ в зал ожидания
    }

    // НОВЫЕ методы 
    public void AccessLounge()                     // Доступ в зал ожидания (НОВЫЙ МЕТОД)
    {
        if (HasLoungeAccess)
        {
            Console.WriteLine($"Доступ в бизнес-зал предоставлен для {HolderName}");
        }
    }
     // Реализация интерфейса IInsurable
    public bool HasInsurance { get; set; }//Имеет страховку
    
    public void ActivateInsurance()                //  Из интерфейса
    {
        HasInsurance = true;
        Console.WriteLine("Страхование путешествий активировано");
    }
//Метод Pay() должен быть переопределен для добавления информации о получении бонусных миль при оплате покупки. 
    
    public override void Pay(decimal amound)
    {
        if (amound <= Balance)
            {
                Balance -=amound;

               // Начисление кэшбэка (из интерфейса)
            decimal cashback = CalculateCashback(amound);
            Balance += cashback;
                // Начисление миль
            int miles = (int)(amound / 50); // (int) отбрасывает дробную часть (не округляет!) 19.99 → 19   
            BonusMiles += miles;    
                    // int earnedMiles=(int)(amound/2);   //  earned - заработанная    (int) отбрасывает дробную часть (не округляет!) 19.99 → 19        
                    // BonusMiles +=earnedMiles;
                    // Console.WriteLine($" Оплата по карте ({CardNumber}) на сумму {amound}, в результате накопленно милей {BonusMiles}, было полученно {earnedMiles}  ");
                
            Console.WriteLine($"Оплата {amound}{Currency}. Кэшбэк: {cashback}{Currency}, Мили: +{miles}");
            }
        else
            {
                // Вывод сообщения о недостатке средств
                Console.WriteLine("Недостаточно средств.");
            }
    }
    public override string GetInfo()
    {
        // Вызов метода базового класса и добавление дополнительной информации
        return base.GetInfo() + $", Type: Gold, Bonus Miles: {BonusMiles}";
    }
// o CheckBalance(): метод для проверки баланса на карте. 
    public override void CheckBalance()
    {
        Console.WriteLine($" Проверка баланса карты ({CardNumber}), на ней {Balance:C}, а так же {BonusMiles} бонусов  ");       //Balance:C  :С это округление   как в России
    }      
}

// 2. PremiumCard (PremiumCreditCard): Должен содержать дополнительные 
// атрибуты, такие как  Ассистент поддержки (SupportAssistant). 
// Метод CheckBalance() должен быть переопределен для предоставления 
// возможности обратиться за помощью к ассистенту поддержки в случае 
// проблем с балансом.
public class PremiumCreditCard : BonusCreditCard, IInsurable, ICreditLimit //Множественное наследование: Один класс + два интерфейса = множественное наследование поведения
{
    public string SupportAssistant { get; set; }
    public string VIPStatus { get; set; }          //  НОВЫЙ атрибут 
    
    // Реализация ICreditLimit
    public decimal CreditLimit { get; private set; }
    public decimal AvailableCredit => CreditLimit + Balance; //  Отрицательный баланс = кредит

public PremiumCreditCard(string _cardNumber, string _holderName, string _expiryDate, string _supportAssistant)//, decimal _balance = 1000 )
                    :base(_cardNumber, _holderName, _expiryDate, 5.0m) // 5% кэшбэк    _balance  )
    {
        SupportAssistant = _supportAssistant;
        VIPStatus = "Platinum";                    //  Установка нового атрибута
        CreditLimit = 500000;                      //  Установка кредитного лимита
    }

     // НОВЫЕ методы 
    public void IncreaseCreditLimit(decimal amount) //  НОВЫЙ МЕТОД
    {
        CreditLimit += amount;
        Console.WriteLine($"Кредитный лимит увеличен до {CreditLimit}{Currency}");
    }

    // Реализация IInsurable
    public bool HasInsurance { get; set; }
    
    public void ActivateInsurance()
    {
        HasInsurance = true; //Имеет страховку
        Console.WriteLine("Премиальное страхование активировано");
    }
    // public override void CheckBalance()
    // {   
    //     base.CheckBalance();
    //     if (Balance<=0)
    //     { Console.WriteLine($" Ваш баланс {Balance:C}, это скорее всего ошибка. Свяжитесь с {SupportAssistant} ");}
    //     else
    //     {
    //         Console.WriteLine($" Проверка баланса карты ({CardNumber}), на ней {Balance:C}. Если возникли проблемы, свяжитесь с {SupportAssistant} ");       //Balance:C  :С это округление   как в России
    //     }
    // }  
     public override void CheckBalance()
    {
        Console.WriteLine($"Баланс: {Balance}{Currency}, Доступно: {AvailableCredit}{Currency}");
        Console.WriteLine($"VIP статус: {VIPStatus}");  // ✅ Использование нового атрибута
    }

    public override string GetInfo()
    {
        // Вызов метода базового класса и добавление дополнительной информации
        return base.GetInfo() + $", Type: Premium, Support: {SupportAssistant}";
    }
 
        
}

// 3. CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен 
// содержать дополнительные атрибуты, такие как Компания (Company). 
// Метод GetInfo() должен быть переопределен для включения информации о 
// компании в описании карты.
public class CorporateCreditCard : CreditCard
{
    public string Company { get; set; }
    
public CorporateCreditCard(string _cardNumber, string _holderName, string _expiryDate, string _company, decimal _balance = 1000 )
                    :base(_cardNumber, _holderName, _expiryDate, _balance  )
    {
        Company = _company;
    }
 
    public override string GetInfo()
    {
        // Вызов метода базового класса и добавление дополнительной информации
        return base.GetInfo() + $", Type: Corporate, Company: {Company}";
    }
 
        
}
    // CreditCard standardCard = new CreditCard("4111 1111 1111 1111", "Ivan Ivanov", "12/25");
    // GoldCreditCard goldCard = new GoldCreditCard("5500 0000 0000 0004", "Petr Petrov", "05/26");
    // PremiumCreditCard premiumCard = new PremiumCreditCard("3400 0000 0000 009", "Anna Sidorova", "10/27", "Mr. Smith");
    // CorporateCreditCard corporateCard = new CorporateCreditCard("3782 8224 6310 005", "Maria Volkova", "01/28", "XYZ Corp");

    //     // Создание массива карт для демонстрации полиморфизма
    //     CreditCard[] cards = { standardCard, goldCard, premiumCard, corporateCard };

    //     // Цикл по всем картам в массиве
    //     foreach (var card in cards)
    //     {
    //         // Вывод информации о каждой карте
    //         Console.WriteLine(card.GetInfo());
    //         // Проверка баланса карты
    //         card.CheckBalance();
    //         // Выполнение платежа на 200 денежных единиц
    //         card.Pay(200);
    //         // Повторная проверка баланса после платежа
    //         card.CheckBalance();
    //         // Пустая строка для разделения вывода по разным картам
    //         Console.WriteLine();
    //     }

    //     // Демонстрация уникального поведения GoldCard
    //     Console.WriteLine("--- Gold Card Bonus Miles Demo ---");
    //     // Выполнение крупного платежа для накопления миль
    //     goldCard.Pay(500);
    //     // Вывод информации о карте с показом накопленных миль
    //     Console.WriteLine(goldCard.GetInfo());

    //     // Демонстрация уникального поведения PremiumCard
    //     Console.WriteLine("\n--- Premium Card Low Balance Demo ---");
    //     // Выполнение крупного платежа для создания низкого баланса
    //     premiumCard.Pay(980);
    //     // Проверка баланса - сработает предложение помощи
    //     premiumCard.CheckBalance();


// class Program
// {
//     static void Main(string[] args)
//     {
//         Console.WriteLine("=== ДЕМОНСТРАЦИЯ ВСЕХ ТИПОВ НАСЛЕДОВАНИЯ ===");
//         Console.WriteLine();

//         // 1. ПРОСТОЕ НАСЛЕДОВАНИЕ - создаем базовые карты
//         Console.WriteLine("1. ПРОСТОЕ НАСЛЕДОВАНИЕ (CreditCard → BonusCreditCard)");
//         Console.WriteLine("=====================================================");
        
//         //BonusCreditCard basicBonusCard = new BonusCreditCard("1111 2222 3333 4444", "Сергей Петров", "06/25", 2.0m);
//         GoldCreditCard basicBonusCard = new GoldCreditCard("1111 2222 3333 4444", "Сергей Петров", "06/25");
        
//         Console.WriteLine(basicBonusCard.GetInfo());
//         basicBonusCard.CheckBalance();
//         basicBonusCard.Pay(300);
//         basicBonusCard.AddBonusPoints(100);
//         Console.WriteLine();

//         // 2. СЛОЖНОЕ НАСЛЕДОВАНИЕ - многоуровневая иерархия
//         Console.WriteLine("2. СЛОЖНОЕ НАСЛЕДОВАНИЕ (CreditCard → BonusCreditCard → GoldCreditCard)");
//         Console.WriteLine("=======================================================================");
        
//         GoldCreditCard goldCard = new GoldCreditCard("5500 0000 0000 0004", "Анна Сидорова", "05/26");
        
//         Console.WriteLine(goldCard.GetInfo());
//         goldCard.CheckBalance();
        
//         // Демонстрация новых возможностей GoldCard
//         goldCard.AccessLounge();                    // ✅ НОВЫЙ МЕТОД
//         goldCard.ActivateInsurance();               // ✅ ИЗ ИНТЕРФЕЙСА
        
//         // Показываем наследование через несколько уровней
//         goldCard.Pay(500);                          // Использует переопределенный метод
//         goldCard.AddBonusPoints(50);                // Метод из промежуточного класса
//         Console.WriteLine();

//         // 3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ - через интерфейсы
//         Console.WriteLine("3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (класс + 2 интерфейса)");
//         Console.WriteLine("===================================================");
        
//         PremiumCreditCard premiumCard = new PremiumCreditCard("3400 0000 0000 009", 
//                                                             "Иван Иванов", "10/27", 
//                                                             "Мария - персональный ассистент");
        
//         Console.WriteLine(premiumCard.GetInfo());
//         premiumCard.CheckBalance();
        
//         // Демонстрация множественного наследования
//         premiumCard.IncreaseCreditLimit(100000);    // ✅ НОВЫЙ МЕТОД
//         premiumCard.ActivateInsurance();            // ✅ ИЗ ПЕРВОГО ИНТЕРФЕЙСА
//         premiumCard.Pay(1500);                      // Использует кредитный лимит
        
//         Console.WriteLine($"VIP статус: {premiumCard.VIPStatus}"); // ✅ НОВЫЙ АТРИБУТ
//         Console.WriteLine();

//         // 4. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА
//         Console.WriteLine("4. ПОЛИМОРФИЗМ - работа с разными типами через базовый класс");
//         Console.WriteLine("============================================================");
        
//         CreditCard[] allCards = { basicBonusCard, goldCard, premiumCard };
        
//         foreach (var card in allCards)
//         {
//             Console.WriteLine($"--- {card.GetType().Name} ---");
//             Console.WriteLine(card.GetInfo());      // Полиморфизм - каждый класс выводит свою информацию
            
//             // Демонстрация работы с валютой (новый атрибут)
//             Console.WriteLine($"Валюта карты: {card.Currency}");
            
//             card.CheckBalance();                    // Полиморфизм - разное поведение
            
//             // Демонстрация управления состоянием (новые методы)
//             if (card.IsActive)
//             {
//                 card.Pay(200);                      // Полиморфизм - разная логика оплаты
//             }
            
//             Console.WriteLine();
//         }

//         // 5. РАБОТА С ИНТЕРФЕЙСАМИ - демонстрация множественного наследования
//         Console.WriteLine("5. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ");
//         Console.WriteLine("==============================================");
        
//         // Все застрахованные карты (из разных иерархий!)
//         IInsurable[] insuredCards = { goldCard, premiumCard };
        
//         Console.WriteLine("Застрахованные карты:");
//         foreach (var insured in insuredCards)
//         {
//             Console.WriteLine($"- {insured.GetType().Name}: Страхование активно - {insured.HasInsurance}");
            
//             // Активируем страхование если еще не активно
//             if (!insured.HasInsurance)
//             {
//                 insured.ActivateInsurance();
//             }
//         }
//         Console.WriteLine();

//         // 6. ДЕМОНСТРАЦИЯ УПРАВЛЕНИЯ СОСТОЯНИЕМ КАРТ (новые методы)
//         Console.WriteLine("6. УПРАВЛЕНИЕ СОСТОЯНИЕМ КАРТ (новые методы)");
//         Console.WriteLine("============================================");
        
//         // Блокировка и активация карт
//         goldCard.BlockCard();
//         Console.WriteLine($"Карта заблокирована: {!goldCard.IsActive}");
        
//         goldCard.ActivateCard();
//         Console.WriteLine($"Карта активирована: {goldCard.IsActive}");
//         Console.WriteLine();

        
//         // Финальная проверка балансов
//         Console.WriteLine("\nФИНАЛЬНЫЕ БАЛАНСЫ:");
//         foreach (var card in allCards)
//         {
//             card.CheckBalance();
//         }
//     }
// }

Console.WriteLine("=== ДЕМОНСТРАЦИЯ НАСЛЕДОВАНИЯ ===");

// 1. ПРОСТОЕ НАСЛЕДОВАНИЕ
Console.WriteLine("\n1. ПРОСТОЕ НАСЛЕДОВАНИЕ");
CreditCard basicCard = new CreditCard("1111 1111 1111 1111", "Иван Иванов", "12/25");
Console.WriteLine(basicCard.GetInfo());
basicCard.Pay(200);

// 2. СЛОЖНОЕ НАСЛЕДОВАНИЕ (3 уровня)
Console.WriteLine("\n2. СЛОЖНОЕ НАСЛЕДОВАНИЕ");
GoldCreditCard goldCard = new GoldCreditCard("5555 5555 5555 5555", "Анна Сидорова", "06/26");
Console.WriteLine(goldCard.GetInfo());
goldCard.Pay(500); // Использует сложную логику оплаты
goldCard.AccessLounge(); // Новая возможность Gold карты

// 3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ
Console.WriteLine("\n3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ");
PremiumCreditCard premiumCard = new PremiumCreditCard("3333 3333 3333 3333", "Петр Петров", "09/27", "Ассистент");
Console.WriteLine(premiumCard.GetInfo());
premiumCard.IncreaseCreditLimit(50000); // Из интерфейса ICreditLimit
premiumCard.ActivateInsurance(); // Из интерфейса IInsurable

// 4. ПОЛИМОРФИЗМ
Console.WriteLine("\n4. ПОЛИМОРФИЗМ");
CreditCard[] cards = { basicCard, goldCard, premiumCard };

foreach (var card in cards)
{
    Console.WriteLine($"--- {card.GetType().Name} ---");
    Console.WriteLine(card.GetInfo());
    card.CheckBalance();
    Console.WriteLine();
}

Console.WriteLine("=== ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА ===");

```
