Создать базовый класс CreditCard в C#, который будет представлять информацию 
о кредитных картах. На основе этого класса разработать 2-3 производных класса,

должны быть реализованы новые атрибуты и методы, а также переопределены 
некоторые методы базового класса для демонстрации полиморфизма. 
Требования к базовому классу CreditCard: 
• Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок 
действия (ExpiryDate). 
• Методы: 
o GetInfo(): метод для получения информации о кредитной карте в виде 
строки. 
o Pay(): метод для оплаты покупки с использованием карты. 
o CheckBalance(): метод для проверки баланса на карте. 
Требования к производным классам: 
1. GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты, 
такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть 
переопределен для добавления информации о получении бонусных миль 
при оплате покупки. 
2. PremiumCard (PremiumCreditCard): Должен содержать дополнительные 
атрибуты, 
такие 
как 
Ассистент 
поддержки

Метод CheckBalance() должен быть переопределен для предоставления 
возможности обратиться за помощью к ассистенту поддержки в случае 
проблем с балансом. 
3. CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен 
содержать дополнительные атрибуты, такие как Компания (Company). 
Метод GetInfo() должен быть переопределен для включения информации о 
компании в описании карты.

```
// дипсик и я 

// 1. Добавим валидацию в базовый класс
public class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    
    // 2. Добавим защиту от отрицательного баланса
blic decimal     puBalance 
    { 
        get => _balance;
        protected set => _balance = value >= 0 ? value 
                : throw new ArgumentException("Баланс не может быть отрицательным");
    }
    private decimal _balance;
public CreditCard(string _cardNumber, string _holderName, string _expiryDate, decimal _balance = 1000 )
{
    CardNumber = _cardNumber;
    HolderName = _holderName;
    ExpiryDate = _expiryDate;
    Balance = _balance;
}
public static int TotalCard{ get; private set; }

    public CreditCard()
    {
        TotalCard++;
    }
    public static void Info()
    {
        Console.WriteLine($"Всего кредитных карт {TotalCard}");
    }

// • Методы:
    public virtual string GetInfo()
    {
        return $"Номер карты ({CardNumber}), Холдера {HolderName}, Срок действия: {ExpiryDate}";
    } 

    // 3. Улучшим метод Pay с обработкой исключений
    public virtual void Pay(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма оплаты должна быть положительной");
            
        if (amount > Balance)
        {
            Console.WriteLine($"Ошибка: Недостаточно средств. Баланс: {Balance:C}, Требуется: {amount:C}");
            return;
        }
        
        Balance -= amount;
        Console.WriteLine($"Успешная оплата {amount:C} с карты {MaskCardNumber(CardNumber)}");
    }
    
    public virtual void CheckBalance()
    {
        Console.WriteLine($" Проверка баланса карты ({CardNumber}), на ней {Balance:C}  ");       //Balance:C  :С это округление   как в России
    } 

    // 4. Вспомогательный метод для маскировки номера карты
    protected string MaskCardNumber(string cardNumber)
    {
        if (cardNumber.Length < 4) return cardNumber;
        return "**** **** **** " + cardNumber.Substring(cardNumber.Length - 4);
    }
}

// 5. Добавим интерфейсы для множественного наследования
public interface IBonusProgram
{
    int BonusPoints { get; }
    void AddBonusPoints(decimal amount);
}

public interface ICustomerSupport
{
    string SupportContact { get; }
    void ContactSupport(string issue);
}

// 6. Улучшенный GoldCreditCard с интерфейсом
public class GoldCreditCard : CreditCard, IBonusProgram
{
    public int BonusPoints { get; private set; }
    public decimal CashbackRate { get; set; } = 0.05m; // 5% кэшбэк
    
    public override void Pay(decimal amount)
    {
        if (base.Pay(amount)) // Если оплата успешна
        {
            AddBonusPoints(amount);
            AddCashback(amount);
        }
    }
    
    public void AddBonusPoints(decimal amount)
    {
        int points = (int)(amount / 10);
        BonusPoints += points;
        Console.WriteLine($"Начислено {points} бонусных баллов");
    }
    
    private void AddCashback(decimal amount)
    {
        decimal cashback = amount * CashbackRate;
        Balance += cashback;
        Console.WriteLine($"Начислен кэшбэк {cashback:C}");
    }
}

// 7. Улучшенный PremiumCreditCard
public class PremiumCreditCard : CreditCard, ICustomerSupport
{
    public string SupportContact { get; set; }
    public bool HasTravelInsurance { get; set; } = true;
    
    public void ContactSupport(string issue)
    {
        Console.WriteLine($"Обращение в поддержку: {issue}");
        Console.WriteLine($"Контакты: {SupportContact}");
    }
    
    public void ActivateTravelInsurance()
    {
        if (HasTravelInsurance)
            Console.WriteLine("Туристическая страховка активирована");
    }
}
```
