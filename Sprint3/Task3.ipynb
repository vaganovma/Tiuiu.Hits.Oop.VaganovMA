```

public interface ICashbackable // Интерфейс для кэшбэка
{
    decimal CashbackPercent { get; }
    decimal CalculateCashback(decimal purchaseAmount);
}

public interface IInsurable // Страхуемый   // Интерфейс для страхования
{
    bool HasInsurance { get; set; }
    void ActivateInsurance();      
}

public interface ICreditLimit  // Интерфейс для кредитного лимита
{
    decimal CreditLimit { get; }
    decimal AvailableCredit { get; }
    void IncreaseCreditLimit(decimal amount);        
}

public abstract class BonusCreditCard : CreditCard, ICashbackable   //Простое наследование: Один базовый класс + один интерфейс
{ 
    public int BonusPoints { get; protected set; }                                         
    public decimal CashbackPercent { get; protected set; }                                    

    public BonusCreditCard(string _cardNumber, string _holderName, string _expiryDate, decimal _cashbackPercent)
        : base(_cardNumber, _holderName, _expiryDate)
    {
        CashbackPercent = _cashbackPercent;
        BonusPoints = 0;
    }                                     
    public virtual void AddBonusPoints(int points)
    { 
        BonusPoints += points;
        Console.WriteLine($"Начислено {points} бонусных баллов. Всего: {BonusPoints}");   
    }                                
    public virtual decimal CalculateCashback(decimal purchaseAmount)
    { 
        return purchaseAmount * CashbackPercent / 100;
    }  
}      

// GENERIC класс для работы с коллекцией карт
public class CardCollection<T> where T : CreditCard
{
    private List<T> _cards = new List<T>();

    public void AddCard(T card)
    {
        _cards.Add(card);
        Console.WriteLine($" Добавлена карта: {card.HolderName}");
    }

    public void RemoveCard(T card)
    {
        if (_cards.Remove(card))
        {
            Console.WriteLine($" Удалена карта: {card.HolderName}");
        }
    }

    public void ProcessAllCards(Action<T> action)
    {
        Console.WriteLine($"\n Обработка {_cards.Count} карт:");
        foreach (var card in _cards)
        {
            action(card);
        }
    }

    public List<T> FindCards(Predicate<T> predicate)
    {
        return _cards.FindAll(predicate);
    }

    public void DisplayAllCards()
    {
        Console.WriteLine($"\n📋 Коллекция карт ({typeof(T).Name}):");
        foreach (var card in _cards)
        {
            Console.WriteLine($"   - {card.GetInfo()}");
        }
    }

    public int Count => _cards.Count;
}

//  Базовому классу CreditCard: • Атрибуты: Номер карты (CardNumber), Холдера (HolderName), Срок 
// действия (ExpiryDate).
public class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    public decimal Balance { get; protected set; }
    // НОВЫЕ атрибуты 
    public string Currency { get; set; }           //  Валюта
    public bool IsActive { get; set; }       
     // НОВЫЙ атрибут 
    public DateTime IssueDate { get; set; }        //  Дата выпуска
    public string BankName { get; set; }    

public CreditCard(string _cardNumber, string _holderName, string _expiryDate, decimal _balance = 1000, string _currency = "RUB" )
{
    CardNumber = _cardNumber;
    HolderName = _holderName;
    ExpiryDate = _expiryDate;
    Balance = _balance;
    Currency = _currency;     //Валюта         
    IsActive = true;                           
    IssueDate = DateTime.Now; // Дата выпуска
    BankName = "Default Bank"; 
}
public static int TotalCard{ get; private set; }

    public CreditCard()
    {
        TotalCard++;
    }
    public static void Info()
    {
        Console.WriteLine($"Всего кредитных карт {TotalCard}");
    }
// • Методы:  
    public virtual void DisplayCardAge() //Возраст карты дисплея     НОВЫЙ метод
    {
        var age = DateTime.Now - IssueDate;
        Console.WriteLine($"Карта выпущена {age.Days} дней назад");
    }
   
    public virtual void ChangeCurrency(string newCurrency) // Изменить валюту   НОВЫЙ метод
    {
        Currency = newCurrency;
        Console.WriteLine($"Валюта карты изменена на {newCurrency}");
    }

    public virtual void ActivateCard()             
    {
        IsActive = true;
        Console.WriteLine($"Карта {CardNumber} активирована");
    }

    public virtual void BlockCard()                 
    {
        IsActive = false;
        Console.WriteLine($"Карта {CardNumber} заблокирована");
    }
// o GetInfo(): метод для получения информации о кредитной карте в виде строки.
    public virtual string GetInfo()
    {
        return $"Номер карты ({CardNumber}), Владелец холдера: {HolderName}, Срок действия: {ExpiryDate}, Банк: {BankName}";
    }
 
// o Pay(): метод для оплаты покупки с использованием карты. 
      // ПЕРЕГРУЗКА методов Pay
    public virtual void Pay(decimal amount)
    {
        Console.WriteLine($"Попытка оплаты на сумму {amount:C} с карты ({CardNumber})");
    }

    public virtual void Pay(decimal amount, string currency)
    {
        Console.WriteLine($"Попытка оплаты на сумму {amount} {currency} с карты ({CardNumber})");
    }

    public virtual void Pay(decimal amount, string description, string merchant)//сумма, описание, продавец
    {
        Console.WriteLine($"Оплата '{description}' у {merchant} на сумму {amount:C}");
    }

// o CheckBalance(): метод для проверки баланса на карте. 
    // public virtual void CheckBalance()
    // {
    //     Console.WriteLine($" Проверка баланса карты ({CardNumber}), на ней {Balance:C}  ");       //Balance:C  :С это округление   как в России
    // } 
    // ПЕРЕГРУЗКА методов CheckBalance
    public virtual void CheckBalance()
    {
        Console.WriteLine($"Баланс карты ({CardNumber}): {Balance:C}");
    }

    public virtual void CheckBalance(bool detailed)
    {
        if (detailed)
        {
            Console.WriteLine($"Детальный баланс:");
            Console.WriteLine($"  Карта: {CardNumber}");
            Console.WriteLine($"  Баланс: {Balance}{Currency}");
            Console.WriteLine($"  Валюта: {Currency}");
            Console.WriteLine($"  Статус: {(IsActive ? "Активна" : "Заблокирована")}");
        }
        else
        {
            CheckBalance();
        }
    }
}




// Требования к производным классам: 
// 1. GoldCard (GoldCreditCard): Должен содержать дополнительные атрибуты, 
// такие как Бесплатные бонусные мили (BonusMiles). Метод Pay() должен быть 
// переопределен для добавления информации о получении бонусных миль 
// при оплате покупки. 
public class GoldCreditCard : BonusCreditCard, IInsurable //Сложное наследование: → CreditCard → BonusCreditCard → GoldCreditCard (3 уровня иерархии)
{
    public int BonusMiles { get; set; }
    public bool HasLoungeAccess { get; set; }      //Имеет доступ в зал ожидания 
    // НОВЫЙ атрибут
    public int LoungeVisits { get; private set; } //Посещения зала
    
public GoldCreditCard(string _cardNumber, string _holderName, string _expiryDate )// ((, decimal _balance = 1000 )  _balance  )
                    :base(_cardNumber, _holderName, _expiryDate, 3.0m) 
    {
        BonusMiles = 1000;                         // Приветственные мили
        HasLoungeAccess = true;                    // (Новый атрибут) Имеет доступ в зал ожидания
        LoungeVisits = 0; // НОВЫЙ атрибут
    }
    // НОВЫЙ метод
    public void DisplayMilesInfo()
    {
        Console.WriteLine($"Накоплено миль: {BonusMiles}");
        Console.WriteLine($"Примерная стоимость: {BonusMiles * 0.5m} руб");
    }
    
    public void AccessLounge()                     // Доступ в зал ожидания (НОВЫЙ МЕТОД)
    {
        if (HasLoungeAccess)
        {
            LoungeVisits++; // НОВЫЙ атрибут
            Console.WriteLine($"Доступ в бизнес-зал предоставлен для {HolderName}");
            Console.WriteLine($"Посещений в этом месяце: {LoungeVisits}");
        }
    }
     // Реализация интерфейса IInsurable
    public bool HasInsurance { get; set; }//Имеет страховку
    
    public void ActivateInsurance()                //  Из интерфейса
    {
        HasInsurance = true;
        Console.WriteLine("Страхование путешествий активировано");
    }
//Метод Pay() должен быть переопределен для добавления информации о получении бонусных миль при оплате покупки. 
    
    // ПЕРЕОПРЕДЕЛЕНИЕ (перекрытие) метода
    public override void Pay(decimal amount)
    {
        if (!IsActive)
        {
            Console.WriteLine("Ошибка: карта не активна");
            return;
        }

        if (amount <= Balance)
        {
            Balance -= amount;
            
            decimal cashback = CalculateCashback(amount); // Начисление кэшбэка (из интерфейса)
            Balance += cashback;
            
            int miles = (int)(amount / 50); // Начисление миль   // (int) отбрасывает дробную часть (не округляет!) 19.99 → 19  
            BonusMiles += miles;
            
            Console.WriteLine($" Gold оплата: {amount}{Currency}");
            Console.WriteLine($"   Кэшбэк: +{cashback}{Currency}, Мили: +{miles}");
        }
        else
        {
            // Вывод сообщения о недостатке средств
            Console.WriteLine("Недостаточно средств.");
        }
    }
    // ПЕРЕГРУЗКА метода Pay для Gold карты
    public void Pay(decimal amount, bool useMiles)
    {
        if (useMiles && BonusMiles >= 1000)
        {
            int milesToUse = (int)(amount * 2); // 1 миля = 0.5 рубля
            if (milesToUse <= BonusMiles)
            {
                BonusMiles -= milesToUse;
                Console.WriteLine($"  Оплата милями: {amount}{Currency}");
                Console.WriteLine($"   Списано миль: {milesToUse}");
            }
            else
            {
                Pay(amount); // Используем обычную оплату
            }
        }
        else
        {
            Pay(amount);
        }
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Type: Gold, Мили: {BonusMiles}, Лаунж: {(HasLoungeAccess ? "Да" : "Нет")}";
    }
// o CheckBalance(): метод для проверки баланса на карте. 
    // public override void CheckBalance()
    // {
    //     Console.WriteLine($" Проверка баланса карты ({CardNumber}), на ней {Balance:C}, а так же {BonusMiles} бонусов  ");       //Balance:C  :С это округление   как в России
    // }      
    public override void CheckBalance()
    {
        Console.WriteLine($"   Gold баланс:");
        Console.WriteLine($"   Баланс: {Balance}{Currency}");
        Console.WriteLine($"   Накоплено миль: {BonusMiles}");
        Console.WriteLine($"   Кэшбэк: {CashbackPercent}%");
    }
}

// 2. PremiumCard (PremiumCreditCard): Должен содержать дополнительные 
// атрибуты, такие как  Ассистент поддержки (SupportAssistant). 
// Метод CheckBalance() должен быть переопределен для предоставления 
// возможности обратиться за помощью к ассистенту поддержки в случае 
// проблем с балансом.
public class PremiumCreditCard : BonusCreditCard, IInsurable, ICreditLimit //Множественное наследование: Один класс + два интерфейса = множественное наследование поведения
{
    public string SupportAssistant { get; set; }
    public string VIPStatus { get; set; }          //  НОВЫЙ атрибут 
    // Реализация ICreditLimit
    public decimal CreditLimit { get; private set; }
    public decimal AvailableCredit => CreditLimit + Balance; //  Отрицательный баланс = кредит
    //public decimal AvailableCredit => Balance >= 0 ? CreditLimit + Balance : CreditLimit - Math.Abs(Balance);999999999999
    // НОВЫЙ атрибут
    public decimal ConciergeBudget { get; private set; }//консьерж-бюджет

public PremiumCreditCard(string _cardNumber, string _holderName, string _expiryDate, string _supportAssistant)
                    :base(_cardNumber, _holderName, _expiryDate, 5.0m) // 5% кэшбэк    
    {
        SupportAssistant = _supportAssistant;
        VIPStatus = "Platinum";                    
        CreditLimit = 500000;                      // кредитного лимита
        ConciergeBudget = 10000;                   // НОВЫЙ атрибут
    }

     // НОВЫЙ метод
    public void UseConciergeService(string service, decimal cost)//Воспользуйтесь услугами консьержа
    {
        if (ConciergeBudget >= cost)
        {
            ConciergeBudget -= cost;
            Console.WriteLine($" Услуга консьержа: {service}");
            Console.WriteLine($"   Стоимость: {cost}{Currency}, Остаток бюджета: {ConciergeBudget}{Currency}");
        }
        else
        {
            Console.WriteLine($" Недостаточно бюджета для услуги: {service}");
        }
    }
 
    public void IncreaseCreditLimit(decimal amount) //  Увеличить кредитный лимит
    {
        CreditLimit += amount;
        Console.WriteLine($"Кредитный лимит увеличен до {CreditLimit}{Currency}");
    }
    // ПЕРЕОПРЕДЕЛЕНИЕ метода Pay
    public override void Pay(decimal amount)
    {
        if (!IsActive)
        {
            Console.WriteLine("Ошибка: карта не активна");
            return;
        }

        decimal available = AvailableCredit;
        if (amount <= available)
        {
            Balance -= amount;
            decimal cashback = CalculateCashback(amount);
            Balance += cashback;
            
            Console.WriteLine($"   Premium оплата: {amount}{Currency}");
            Console.WriteLine($"   Кэшбэк: +{cashback}{Currency}");
            Console.WriteLine($"   Доступно: {AvailableCredit}{Currency}");
        }
        else
        {
            Console.WriteLine(" Превышен кредитный лимит");
        }
    }

    // Реализация IInsurable
    public bool HasInsurance { get; set; }
    
    public void ActivateInsurance()
    {
        HasInsurance = true; //Имеет страховку
        Console.WriteLine("Премиальное страхование активировано");
    }
    // public override void CheckBalance()
    // {   
    //     base.CheckBalance();
    //     if (Balance<=0)
    //     { Console.WriteLine($" Ваш баланс {Balance:C}, это скорее всего ошибка. Свяжитесь с {SupportAssistant} ");}
    //     else
    //     {
    //         Console.WriteLine($" Проверка баланса карты ({CardNumber}), на ней {Balance:C}. Если возникли проблемы, свяжитесь с {SupportAssistant} ");       //Balance:C  :С это округление   как в России
    //     }
    // }  
    //  public override void CheckBalance()
    // {
    //     Console.WriteLine($"Баланс: {Balance}{Currency}, Доступно: {AvailableCredit}{Currency}");
    //     Console.WriteLine($"VIP статус: {VIPStatus}");  //  Использование нового атрибута
    // }

    
    public override void CheckBalance()
    {
        Console.WriteLine($"   Premium баланс:");
        Console.WriteLine($"   Баланс: {Balance}{Currency}");
        Console.WriteLine($"   Кредитный лимит: {CreditLimit}{Currency}");
        Console.WriteLine($"   Доступно: {AvailableCredit}{Currency}");
        Console.WriteLine($"   VIP: {VIPStatus}, Бюджет консьержа: {ConciergeBudget}{Currency}");
    }

    public override string GetInfo()
    {   // Вызов метода базового класса и добавление дополнительной информации
        return base.GetInfo() + $", Type: Premium, Лимит: {CreditLimit}{Currency}, Ассистент: {SupportAssistant}";
    }
 
        
}

// 3. CorporateCard (CorporateCreditCard) (если требуется третий класс): Должен 
// содержать дополнительные атрибуты, такие как Компания (Company). 
// Метод GetInfo() должен быть переопределен для включения информации о 
// компании в описании карты.
public class CorporateCreditCard : CreditCard
{
    public string Company { get; set; }
    
public CorporateCreditCard(string _cardNumber, string _holderName, string _expiryDate, string _company, decimal _balance = 1000 )
                    :base(_cardNumber, _holderName, _expiryDate, _balance  )
    {
        Company = _company;
    }
 
    public override string GetInfo()
    {
        // Вызов метода базового класса и добавление дополнительной информации
        return base.GetInfo() + $", Type: Corporate, Company: {Company}";
    }
 
        
}
  {
      // CreditCard standardCard = new CreditCard("4111 1111 1111 1111", "Ivan Ivanov", "12/25");
    // GoldCreditCard goldCard = new GoldCreditCard("5500 0000 0000 0004", "Petr Petrov", "05/26");
    // PremiumCreditCard premiumCard = new PremiumCreditCard("3400 0000 0000 009", "Anna Sidorova", "10/27", "Mr. Smith");
    // CorporateCreditCard corporateCard = new CorporateCreditCard("3782 8224 6310 005", "Maria Volkova", "01/28", "XYZ Corp");

    //     // Создание массива карт для демонстрации полиморфизма
    //     CreditCard[] cards = { standardCard, goldCard, premiumCard, corporateCard };

    //     // Цикл по всем картам в массиве
    //     foreach (var card in cards)
    //     {
    //         // Вывод информации о каждой карте
    //         Console.WriteLine(card.GetInfo());
    //         // Проверка баланса карты
    //         card.CheckBalance();
    //         // Выполнение платежа на 200 денежных единиц
    //         card.Pay(200);
    //         // Повторная проверка баланса после платежа
    //         card.CheckBalance();
    //         // Пустая строка для разделения вывода по разным картам
    //         Console.WriteLine();
    //     }

    //     // Демонстрация уникального поведения GoldCard
    //     Console.WriteLine("--- Gold Card Bonus Miles Demo ---");
    //     // Выполнение крупного платежа для накопления миль
    //     goldCard.Pay(500);
    //     // Вывод информации о карте с показом накопленных миль
    //     Console.WriteLine(goldCard.GetInfo());

    //     // Демонстрация уникального поведения PremiumCard
    //     Console.WriteLine("\n--- Premium Card Low Balance Demo ---");
    //     // Выполнение крупного платежа для создания низкого баланса
    //     premiumCard.Pay(980);
    //     // Проверка баланса - сработает предложение помощи
    //     premiumCard.CheckBalance();
  }
    

Console.WriteLine("=== ДЕМОНСТРАЦИЯ: ПОЛИМОРФИЗМ + ПЕРЕГРУЗКА + GENERICS ===");

// Создаем generic коллекции
var goldCollection = new CardCollection<GoldCreditCard>();
var premiumCollection = new CardCollection<PremiumCreditCard>();
var allCardsCollection = new CardCollection<CreditCard>();

// Создаем карты
var goldCard = new GoldCreditCard("5555 5555 5555 5555", "Анна Сидорова", "06/26");
var premiumCard = new PremiumCreditCard("3333 3333 3333 3333", "Петр Петров", "09/27", "Мария-ассистент");

// Добавляем в коллекции
goldCollection.AddCard(goldCard);
premiumCollection.AddCard(premiumCard);
allCardsCollection.AddCard(goldCard);
allCardsCollection.AddCard(premiumCard);

// 1. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА
Console.WriteLine("\n1.  ПОЛИМОРФИЗМ (переопределение методов)");
CreditCard[] cards = { goldCard, premiumCard };
foreach (var card in cards)
{
    Console.WriteLine($"\n--- {card.GetType().Name} ---");
    card.Pay(300); // Полиморфный вызов - разное поведение
    card.CheckBalance();
}

// 2. ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ
Console.WriteLine("\n2.  ПЕРЕГРУЗКА МЕТОДОВ");
goldCard.Pay(200); // Базовая версия
goldCard.Pay(150, "EUR"); // С валютой
goldCard.Pay(100, "Ресторан", "Il Patio"); // С описанием
goldCard.Pay(500, true); // С использованием миль (перегрузка в Gold классе)

Console.WriteLine();
goldCard.CheckBalance(); // Простая версия
goldCard.CheckBalance(true); // Детальная версия

// 3. ДЕМОНСТРАЦИЯ GENERICS
Console.WriteLine("\n3.  GENERICS КОЛЛЕКЦИИ");
goldCollection.DisplayAllCards();
premiumCollection.DisplayAllCards();

Console.WriteLine($"\nВсего карт в коллекции: {allCardsCollection.Count}");

// Обработка через generic методы
goldCollection.ProcessAllCards(card => 
{
    card.AccessLounge();
    card.DisplayMilesInfo();
});

// 4. НОВЫЕ МЕТОДЫ И АТРИБУТЫ
Console.WriteLine("\n4.  НОВЫЕ ВОЗМОЖНОСТИ");
goldCard.DisplayCardAge();
goldCard.DisplayMilesInfo();

premiumCard.UseConciergeService("Бронь отеля", 2000);
premiumCard.UseConciergeService("Заказ билетов", 1500);

// 5. РАБОТА С ИНТЕРФЕЙСАМИ
Console.WriteLine("\n5.  ИНТЕРФЕЙСЫ");
IInsurable[] insuredCards = { goldCard, premiumCard };
foreach (var insured in insuredCards)
{
    if (!insured.HasInsurance)
    {
        insured.ActivateInsurance();
    }
}

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА ===");






```
